````python
def is_leap_year(year):

  ans = False
  if year % 400 == 0:
    ans = True
  elif year % 100 == 0:
    ans = False
  elif year % 4 == 0:
    ans = True
  else:
    ans = False
  return ans
# year = int(input("請輸入西元年: "))
# print(is_leap_year(year))
````

````python
# 判斷日期是否合理
def is_date_valid(date_check):
  # date_check = int(input("請輸入西元年月日: "))
  date_list = {
      1: 31,
      2: 28,
      3: 31,
      4: 30,
      5: 31,
      6: 30,
      7: 31,
      8: 31,
      9: 30,
      10: 31,
      11: 30,
      12: 31
  }
  # print(date_check[1])
  day = date_check % 100
  month = (date_check // 100) % 100
  year = (date_check // 10000)
  ans = False

  if year > 0:
    if month > 12 or month < 1:
      # print("月份不合理")
      ans = False
    else:
      if is_leap_year(year) == True and month == 2:
        date_list[month] = date_list[month] + 1
      if 0 < day <= date_list[month]:
        # print(year, "/", month, "/", day, "\n日期合理")
        ans = True
      else:
        # print("日期不合理")
        ans = False
  else:
    # print("年份不合理")
    ans = False
  return ans
# print(is_date_valid())
````

````python
def fully_squared_1(start=4499, max_count=10):
  # start = 4499
  date = start ** 2
  count = 0
  ans = []
  while True:
    date = start ** 2
    if is_date_valid(date) == True:
      count = count + 1
      ans.append(date)
    start = start + 1
    if count == max_count:
      break
  return ans
fully_squared_1(123, 10)
````

````python
import datetime as dt

def fully_squared_2(base_y=2024, base_m=10, base_d=1, max_count=10):
  ans = []
  base = dt.datetime(base_y, base_m ,base_d)
  pat = "%Y%m%d"
  count = 0
  delta = dt.timedelta(days=1)
  while True:
    base_str = base.strftime(pat)
    base_int = int(base_str)
    base_root = base_int ** 0.5
    dif = abs(round(base_root) - base_root  )
    # round 四捨五入 - 原始小於門檻值 就認為一樣
    if dif < 10 ** -6:
      # print(base)
      ans.append(base_int)
      count = count + 1
    base = base + delta
    if count == max_count:
      break
  return ans
fully_squared_2()
````

````python
#比較兩想法花的時間

import time
try_count = 40

start = time.time()
fully_squared_1(max_count=try_count)
end = time.time()
dif1 = end - start
print("想法一花了", dif1, "秒")

start = time.time()
fully_squared_2(max_count=try_count)
end = time.time()
dif2 = end - start
print("想法二花了", dif2, "秒")

ratio = dif2 / dif1
print("想法1比想法2快", ratio, "倍")
````
