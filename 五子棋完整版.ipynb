```python
# 五子棋
# 1. 先畫棋盤
# 2. 落子(落到最相鄰的位置)
# 3. 判斷落子是否五子

import pygame as pg

def check_win(board, player_i, player_j):
    count = 0
    same_target = 4
    player_check = board[player_i][player_j]
    # 向右
    offset  = 1
    while True:
        new_i, new_j = player_i, player_j + offset
        if 0 <= new_i <len(board) and 0 <= new_j < len(board[0]):
            if board[new_i][new_j] == player_check:
                count = count + 1
            else:
                # 遇到不同色
                break
        else:
            # 撞牆
            break
        if count == same_target:
            return True
        offset = offset + 1

    # 向左
    offset = 1
    while True:
        new_i, new_j = player_i, player_j - offset
        if 0 <= new_i < len(board) and 0 <= new_j < len(board[0]):
            # print(board[0], board[1])
            if board[new_i][new_j] == player_check:
                count = count + 1
            else:
                # 遇到不同色
                break
        else:
            # 撞牆
            break
        if count == same_target:
            return True
        offset = offset + 1


    # 向上
    offset  = 1
    while True:
        new_i, new_j = player_i + offset, player_j
        if 0 <= new_i <len(board) and 0 <= new_j < len(board[0]):
            if board[new_i][new_j] == player_check:
                count = count + 1
            else:
                # 遇到不同色
                break
        else:
            # 撞牆
            break
        if count == same_target:
            return True
        offset = offset + 1

    # 向下
    count = 0
    offset = 1
    while True:
        new_i, new_j = player_i - offset, player_j
        if 0 <= new_i < len(board) and 0 <= new_j < len(board[0]):
            # print(board[0], board[1])
            if board[new_i][new_j] == player_check:
                count = count + 1
            else:
                # 遇到不同色
                break
        else:
            # 撞牆
            break
        if count == same_target:
            return True
        offset = offset + 1

    # 向右上
    count = 0
    offset = 1
    while True:
        new_i, new_j = player_i - offset, player_j + offset
        if 0 <= new_i < len(board) and 0 <= new_j < len(board[0]):
            if board[new_i][new_j] == player_check:
                count = count + 1
            else:
                # 遇到不同色
                break
        else:
            # 撞牆
            break
        if count == same_target:
            return True
        offset = offset + 1
    # 向左下
    offset = 1
    while True:
        new_i, new_j = player_i + offset, player_j - offset
        if 0 <= new_i < len(board) and 0 <= new_j < len(board[0]):
            if board[new_i][new_j] == player_check:
                count = count + 1
            else:
                # 遇到不同色
                break
        else:
            # 撞牆
            break
        if count == same_target:
            return True
        offset = offset + 1
    # 向左上
    count = 0
    offset = 1
    while True:
        new_i, new_j = player_i - offset, player_j - offset
        if 0 <= new_i < len(board) and 0 <= new_j < len(board[0]):
            if board[new_i][new_j] == player_check:
                count = count + 1
            else:
                # 遇到不同色
                break
        else:
            # 撞牆
            break
        if count == same_target:
            return True
        offset = offset + 1
    # 向右下
    offset = 1
    while True:
        new_i, new_j = player_i + offset, player_j + offset
        if 0 <= new_i < len(board) and 0 <= new_j < len(board[0]):
            # print(board[0], board[1])
            if board[new_i][new_j] == player_check:
                count = count + 1
            else:
                # 遇到不同色
                break
        else:
            # 撞牆
            break
        if count == same_target:
            return True
        offset = offset + 1
    return False
    # for i in range(same_target):
    #     if  board[player_i][player_j+(i+1)] == player_check:
    #         count = count + 1
    #     else:
    #         break
    #
    # if count == same_target:
    #     return True
    #


# pygame 初始化
pg.init()
# 設定視窗
width, height = 600, 600
# 產生視窗
screen = pg.display.set_mode([width, height])
# 設定遊戲練習
pg.display.set_caption("五子棋")
# 建立畫布bg
bg = pg.Surface(screen.get_size())
# 建立棋盤線數
board_line = 15

# 把畫布填滿某顏色
bg.fill([199, 167, 82])

# pg.draw.line(bg, [100, 100, 100], [40, 40], [600, 40], 2)
x_inter, y_inter = width / (board_line + 1), height / (board_line + 1)
board_color = [100, 100, 100]
i = 0
while i < board_line:
    # row
    pg.draw.line(bg,
                 board_color,
                 [x_inter, y_inter+i*y_inter],
                 [width-x_inter, y_inter+i*y_inter],
                 2)
    # colum
    pg.draw.line(bg,
                 board_color,
                 [x_inter+i*x_inter, y_inter],
                 [x_inter+i*x_inter, height-y_inter],
                 2)
    i = i + 1



# i = 0
# while i < 15:
#     pg.draw.line(bg, [50,50,50], [40, 40*i+40], [600, 40*i+40], 2)
#     i = i + 1

# 把某個圖層貼到上一層
screen.blit(bg, dest=[0, 0])
# 對畫面進行更新
pg.display.update()
player1_symbol, player2_symbol, player_unset = 1, 2, 0
game_board = [[player_unset] * board_line for i in range(board_line)]
print(game_board)

game_round = 0
running = True
while running:
    for event in pg.event.get():

        # 當點擊滑鼠後放開
        if event.type == pg.MOUSEBUTTONUP:
            # 取得放開當下座標
            x, y = pg.mouse.get_pos()
            # 將座標縮小: 0 40 80 120-> 0 1 2 3...
            x_scale, y_scale = x / x_inter, y / y_inter
            # 四捨五入取相鄰位置(看到底靠近左邊還右邊)
            x_scale_round, y_scale_round = round(x_scale), round(y_scale)
            # 放大回原來座標
            x_real, y_real = x_scale_round * x_inter, y_scale_round * y_inter

            board_idx_i, board_idx_j = y_scale_round - 1, x_scale_round - 1
            print(board_idx_i, board_idx_j)

            # 過濾棋盤格以外位置
            if  (1 <= x_scale_round <= board_line) and (1 <= y_scale_round <= board_line):
                if game_board[board_idx_i][board_idx_j] == player_unset:
                    # 棋子半徑(配合長寬不同處理)
                    circle_radius = min(x_inter/2, y_inter/2)
                    # 畫棋子(顏色座標半徑)
                    # 黑子開始, 黑白交替
                    if game_round % 2 == 0:
                        player_color = [0, 0, 0]
                        player_symbol = player1_symbol
                        # game_board[board_idx_i][board_idx_j] = player1_symbol
                    else:
                        player_color = [255, 255, 255]
                        player_symbol = player2_symbol
                        # game_board[board_idx_i][board_idx_j] = player2_symbol
                    game_board[board_idx_i][board_idx_j] = player_symbol
                    pg.draw.circle(bg, player_color, [x_real, y_real], circle_radius)
                    # 把某個圖層貼到上一層
                    screen.blit(bg, dest=[0, 0])
                    # 對畫面進行更新
                    pg.display.update()
                    game_round = game_round + 1
                    if check_win(game_board, board_idx_i, board_idx_j) == True:
                        running = False
            print(game_board[board_idx_i][board_idx_j])
        # 如果收到的事件是按X

        if event.type == pg.QUIT:
            # 迴圈就會變成while False
            running = False
pg.quit()
```
